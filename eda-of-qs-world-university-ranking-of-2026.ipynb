## 🗄️ **About dataset**
This dataset has details of 1501 international universities that are extracted from the QS World University ranking of 2026.

Dataset column names:
- **2026 Rank** - University rank in 2026 QS world university ranking
- **Previous Rank** - University rank in 2025 QS world university ranking
- **Institution Name** - Full name of the university
- **Country/Territory** - The country where the institution is located.
- **Region** - Geographical region (e.g., Europe, Americas, Asia).
- **Size** - Approximate size of the university based on student population (e.g., S, M, L, XL)
- **Focus** - The academic focus of the institution (e.g., CO = Comprehensive, FC = Focused)
- **Research** - Level of research intensity (e.g., VH = Very High)
- **Status** - Type of institution (e.g., Public, Private not-for-profit)
- **AR SCORE** - Academic Reputation score of the university based on global academic survey.
- **AR RANK** - Rank of the university based on AR score
- **ER SCORE** - Employer Reputation score of the university based on employer feedback.
- **ER RANK** - Rank of the university based on ER score.
- **FSR SCORE** - Faculty Student Ratio score which evaluates teaching commitment.
- **FSR RANK** - Rank of the university based on FSR score.
- **CPF SCORE** - Citations Per Faculty score which measures research output and quality.
- **CPF RANK** - Rank of the university based on CPF score.
- **IFR SCORE** - International Faculty Ratio which denotes percentage of international teaching staff in the university.
- **IFR RANK** - Rank of the university based on IFR score.
- **ISR SCORE** - International Student Ratio which denotes percentage of international students.
- **ISR RANK** - Rank of the university based on ISR score.
- **ISD SCORE** - International Student Density which denotes diversity among the international students in the university.
- **ISD RANK** - Rank of the university based on ISD score.
- **IRN SCORE** - International Research Network which denotes global collaboration intensity.
- **IRN RANK** - Rank of the university based on IRN score.
- **EO SCORE** - Employment and Outcome score which denotes how well graduates perform in the job market.
- **EO RANK** - Rank of the university based on EO score.
- **SUS SCORE** - Sustainability score which evaluates environmental and societal impact initiatives.
- **SUS RANK** - Rank of the university based on SUS score.
- **Overall SCORE** - Composite score used to determine the final university rank.

- Identify the employment and outcome of universities based on AR score and research score.
